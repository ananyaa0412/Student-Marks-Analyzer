STUDENTS MARKS ANALYZER


```python
import numpy as np
def marks(avg):
  if avg>=85:
    return "A"
  elif avg>=75:
    return "B"
  elif avg>=50:
    return "C"
  else:
    return "Fail"

  n=int(input("Enter no.of students: "))
  marks=[]

  for i in range(n):
    mark=int(input("Enter marks of student {}:",format(i+1)))
    marks.append(mark)
    arr=np.array(marks)
    total=np.sum(arr)
    average=np.avg(arr)
    highest=np.max(arr)
    lowest=np.min(arr)
    grade=grade(avg)

    print("STUDENT MARKS ANALYZER\n")
    print("Marks:",arr)
    print("Total Marks:",total)
    print("Average Marks:",average)
    print("Highest Marks:",highest)
    print("Lowest Marks:",lowest)
```
